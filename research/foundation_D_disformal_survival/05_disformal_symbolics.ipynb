# Foundation D — Disformal Symbolic Checks

Verify the coupling structures and classify effective metrics
for each toy action.

```python
import sympy as sp
from sympy import sqrt, symbols, simplify, Rational

M_Pl = symbols('M_Pl', positive=True)
H = symbols('H', positive=True)
g_eff = symbols('g_eff', positive=True)
mu = symbols('mu', positive=True)
Z = symbols('Z', positive=True)
b1, b2 = symbols('b_1 b_2')
c1, c2 = symbols('c_1 c_2')
f_a = symbols('f_a', positive=True)  # pseudoscalar decay constant
```

## Toy I: 0⁻ Mode Coupling Structure

Coupling: g_eff (∂_μ B)(ψ̄ γ^μ γ₅ ψ)

On FRW with B = B₀(t): the only nonzero derivative is Ḃ₀.
Effect: chiral energy shift ΔE = ±g_eff Ḃ₀.

```python
# Toy I: Chiral energy shift
# Ḃ₀ ~ H × f_a (for a rolling pseudoscalar with amplitude ~ f_a)
B_dot = H * f_a
Delta_E = g_eff * B_dot

print('Toy I: 0⁻ torsion-fermion coupling')
print('='*50)
print(f'Ḃ₀ ~ H × f_a')
print(f'Chiral energy shift: ΔE = g_eff × Ḃ₀ = {Delta_E}')
print()

# For g_eff ~ 1/M_Pl (gravitational strength) and f_a ~ M_Pl:
Delta_E_num = Delta_E.subs([(g_eff, 1/M_Pl), (f_a, M_Pl)])
print(f'At gravitational coupling: ΔE/H = {simplify(Delta_E_num / H)}')
print('So ΔE ~ H — the energy shift is of order the Hubble scale.')
print()
print('This produces ALP birefringence (rotation of CMB polarization).')
print('NOT a disformal metric modification — standard ALP physics.')
print()
print('CMB rotation angle: Δα ~ g_eff × ΔB')
print(f'  For g_eff = 1/M_Pl, ΔB ~ M_Pl: Δα ~ 1 radian (excluded)')
print(f'  For g_eff << 1/M_Pl: Δα << 1 (allowed, standard ALP constraint)')
```

## Toy II: 0⁺ Mode — Removability Check

Coupling: g_eff (∂_μ φ)(ψ̄ γ^μ ψ)

Field redefinition ψ → exp(-i g_eff φ) ψ removes the coupling.

```python
# Toy II: Removability
print('Toy II: 0⁺ torsion-fermion coupling')
print('='*50)
print()
print('Original coupling: g_eff (∂_μφ)(ψ̄γ^μψ)')
print('Field redefinition: ψ → exp(-i g_eff φ) ψ')
print()
print('After redefinition:')
print('  Kinetic: ψ̄ iγ^μ ∂_μ ψ → ψ̄\' iγ^μ(∂_μ + ig_eff ∂_μφ)ψ\'')
print('         = ψ̄\' iγ^μ ∂_μ ψ\' + ig_eff(∂_μφ)(ψ̄\'γ^μψ\')')
print('  The coupling term CANCELS the original g_eff(∂_μφ)(ψ̄γ^μψ).')
print()
print('  Mass term: m ψ̄ψ → m ψ̄\'ψ\' (invariant under vector phase).')
print()
print('Result: fermion sector is INDEPENDENT of φ.')
print('The 0⁺ coupling is completely gauge-removable.')
print()
print('CLASSIFICATION: TRIVIAL — no physical coupling.')
```

## Toy III: R² Modified GW Speed

```python
# Toy III: Modified GW speed from R² terms
R_FRW = 12 * H**2  # de Sitter approximation

# c²_GW ≈ 1 + (2b₂ - 4b₁) R̄ / M_Pl²
Delta_c2 = (2*b2 - 4*b1) * R_FRW / M_Pl**2

print('Toy III: Modified GW speed')
print('='*50)
print()
print(f'Δc²_GW = (2b₂ - 4b₁) × R̄/M_Pl² = {Delta_c2}')
print()
print(f'On FRW (R ~ 12H²):')
print(f'  Δc² ~ (2b₂ - 4b₁) × 12H²/M_Pl²')
print()
print('Numerically:')
print('  H₀ ~ 10⁻³³ eV, M_Pl ~ 2.4×10¹⁸ GeV = 2.4×10²⁷ eV')
print('  H₀²/M_Pl² ~ (10⁻³³)²/(2.4×10²⁷)² ~ 10⁻¹²⁰/(6×10⁵⁴) ~ 10⁻¹²¹')
print(f'  |Δc²| ~ |2b₂ - 4b₁| × 10⁻¹²⁰')
print()
print('GW170817 constraint: |Δc²| < 10⁻¹⁵')
print('For b_i ~ O(1): |Δc²| ~ 10⁻¹²⁰ << 10⁻¹⁵. Satisfied trivially.')
print()
print('The effect is unobservably small — 10⁵ orders below detection.')
print('CLASSIFICATION: GENERIC_DISFORMAL (Planck-suppressed, not distinctive)')
```

## Toy IV: Non-Minimal Curvature-Fermion → Disformal

```python
# Toy IV: Disformal coefficient from non-minimal coupling
#
# g̃_μν = g_μν + B_dis × (∂_μB)(∂_νB)
# where B_dis = c₁ c₂ / M_Pl⁴
#
# Magnitude of disformal correction:
# |δg̃/g| ~ B_dis × (∂B)² ~ (c₁c₂/M_Pl⁴) × (H f_a)²

B_dis = c1 * c2 / M_Pl**4
dB_sq = (H * f_a)**2  # (∂B)² on FRW
delta_g = B_dis * dB_sq

print('Toy IV: Non-minimal disformal coupling')
print('='*50)
print()
print(f'Disformal coefficient: B_dis = c₁c₂/M_Pl⁴')
print(f'(∂B)² on FRW: ~ (H f_a)²')
print(f'Disformal correction: |δg̃/g| = {delta_g}')
print()

# For f_a ~ M_Pl, c₁ ~ c₂ ~ 1:
delta_g_num = delta_g.subs([(c1, 1), (c2, 1), (f_a, M_Pl)])
print(f'At f_a = M_Pl, c₁ = c₂ = 1:')
print(f'  |δg̃/g| = {simplify(delta_g_num)} = H²/M_Pl²')
print(f'  Numerically: ~ 10⁻¹²²')
print()
print('This is a dimension-8 operator suppressed by M_Pl⁻⁴.')
print('Completely unobservable.')
print()
print('CLASSIFICATION: GENERIC_DISFORMAL (M_Pl⁻⁴ suppressed)')
```

## Non-Metricity Decoupling Verification

```python
# Verify: ω_{(ab)} γ^{ab} = 0
#
# γ^{ab} = ½[γ^a, γ^b] is antisymmetric in a,b
# ω_{(ab)} is symmetric in a,b
# Contraction of symmetric with antisymmetric = 0

print('Non-metricity decoupling from Dirac equation')
print('='*50)
print()
print('Connection in Dirac equation: D_μψ = ∂_μψ + ¼ω_{μab}γ^{ab}ψ')
print()
print('Decompose: ω_{μab} = ω_{μ[ab]} + ω_{μ(ab)}')
print('where ω_{μ(ab)} = Q_{μab}/2 (non-metricity)')
print()
print('γ^{ab} = ½[γ^a, γ^b] satisfies γ^{ab} = -γ^{ba} (antisymmetric)')
print()
print('Therefore:')
print('  ω_{μ(ab)} γ^{ab} = Σ_{a,b} ω_{μ(ab)} γ^{ab}')
print('  = Σ_{a<b} ω_{μ(ab)} γ^{ab} + Σ_{a>b} ω_{μ(ab)} γ^{ab}')
print('  = Σ_{a<b} ω_{μ(ab)} γ^{ab} + Σ_{a<b} ω_{μ(ba)} γ^{ba}')
print('  = Σ_{a<b} ω_{μ(ab)} γ^{ab} + Σ_{a<b} ω_{μ(ab)} (-γ^{ab})')
print('  = 0')
print()
print('QED: Non-metricity drops out of the Dirac equation.')
print('Fermions do not couple to non-metricity in the standard prescription.')
```

## Connection Coupling vs Disformal: Structural Argument

```python
print('Why connection coupling cannot produce disformal structure')
print('='*60)
print()
print('Connection coupling to fermions:')
print('  D_μψ = ∂_μψ + ¼ω_{μab}γ^{ab}ψ')
print('  → γ^μ D_μ ψ = γ^μ∂_μψ + ¼ω_{μab}γ^μγ^{ab}ψ')
print()
print('Structure: one index μ from the connection contracted with γ^μ.')
print('After torsion decomposition: ω → LC + K(torsion)')
print('For propagating torsion B: K_μ ~ ∂_μB × (Lorentz generators)')
print()
print('Result: (∂_μB) × γ^μ × (Lorentz structure) × ψ')
print('This has ONE free Lorentz index (μ) from ∂_μB.')
print()
print('Disformal metric requires:')
print('  g̃_{μν} = g_{μν} + B(φ) ∂_μφ ∂_νφ')
print('  → fermion kinetic: ψ̄ γ^μ g̃_{μν} ∂^ν ψ')
print('  → needs (∂_μφ)(∂_νφ)(ψ̄ γ^μ ∂^ν ψ)')
print('  This has TWO free indices (μ,ν) from ∂_μφ ∂_νφ.')
print()
print('CONNECTION COUPLING: 1 derivative of scalar per vertex')
print('DISFORMAL METRIC: 2 derivatives of scalar per vertex')
print()
print('The connection structure provides ONE ∂φ per coupling.')
print('Getting TWO requires either:')
print('  (a) two connection insertions (loop level → suppressed)')
print('  (b) non-minimal coupling R_{μν}(ψ̄γ^μ∂^νψ) → dim-8, M_Pl⁻⁴')
print()
print('CONCLUSION: Disformal structure from connection coupling is either')
print('absent (tree level) or M_Pl⁻⁴ suppressed (non-minimal coupling).')
```
